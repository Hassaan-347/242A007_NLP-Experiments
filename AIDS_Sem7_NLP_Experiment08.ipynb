```python
# Hassaan Shaikh - 242A007
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import random
```

```python
# Hassaan Shaikh - 242A007
episodes = [
    "Join Penny's world as she engages in various misadventures and gets into fights she wasn't supposed to.",
    "Jeremy's death has stirred The Hooligans to look for the vigilantes who disrupted their operation. When the remainder of Jeremy's men get back to cleaning up, they recognize Penny and her team.",
    "Thomas accidentally sends out Penny's explosives disguised as cookies instead of the real deal. On the other hand, 8 is experiencing a buggy code.",
    "Penny and Walter visit Beats Palace for a night of fun. The Hooligans also happen to be there looking to commit a felony when they spot Penny. A certain Frenchman is keenly interested in Penny.",
    "Thomas and Julie leave their sons under Penny's care before traveling for work. The boys are thoroughly impressed by Penny as she juggles between the responsibility of an aunt and fighting off The Hooligans. However, things get desperate when Lewis gets abducted.",
]
```

```python
# Hassaan Shaikh - 242A007
for i, episode in enumerate(episodes):
    print(f"E{i + 1}: {episode}")
```

```python
# Hassaan Shaikh - 242A007
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(episodes)
print(f"TF-IDF Matrix Shape: {tfidf_matrix.shape}")
print("Doug was here...")
print(f"Feature Names: {vectorizer.get_feature_names_out()}")
```

```python
# Hassaan Shaikh - 242A007
print(tfidf_matrix[random.randint(0, len(episodes) - 1):])
```

```python
# Hassaan Shaikh - 242A007
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out(), index=['E1', 'E2', 'E3', 'E4', 'E5'])
tfidf_df.round(3)
```

```python
# Hassaan Shaikh - 242A007
similarity_matrix = cosine_similarity(tfidf_matrix)
similarity_matrix
```

```python
# Hassaan Shaikh - 242A007
similarity_df = pd.DataFrame(similarity_matrix, columns=['E1', 'E2', 'E3', 'E4', 'E5'], index=['E1', 'E2', 'E3', 'E4', 'E5'])
similarity_df.round(3)
```

```python
# Hassaan Shaikh - 242A007
e1_index = random.randint(0, len(episodes) - 1)
e2_index = random.randint(0, len(episodes) - 1)
while e2_index == e1_index:
    e2_index = random.randint(0, len(episodes) - 1)

similarity = similarity_matrix[e1_index][e2_index]
print(f"Similarity between Episode {e1_index + 1} and Episode {e2_index + 1}: {similarity:.3f}")
```

```python
# Hassaan Shaikh - 242A007
highest_similarity = -1
most_similar_pair = None

for i in range(len(episodes)):
    for j in range(i + 1, len(episodes)):
        similarity = similarity_matrix[i][j]
        if similarity > highest_similarity:
            highest_similarity = similarity
            most_similar_pair = (i, j)

print(f"Most similar pair of episodes: Episode {most_similar_pair[0] + 1} and Episode {most_similar_pair[1] + 1}")
print(f"Similarity: {highest_similarity:.3f}")
```

```python
def vectorized_similarity(texts: list[str], show_internal_data = False):
  vectorized_matrix = vectorizer.fit_transform(texts)
  if show_internal_data:
    print("TF-IDF Vectorizer:")
    print(f"TF-IDF Matrix Shape: {vectorized_matrix.shape}")
    print(f"Feature Names: {vectorizer.get_feature_names_out()}\n")

  return cosine_similarity(vectorized_matrix)
```
